````python
# ============================================================
# 第 2 格：建立資料庫與初始資料
# 提醒：每次執行都要 DROP TABLE IF EXISTS 再重建
# ============================================================
# ============================================================
# 第 2 格：建立資料庫與初始資料
# ============================================================


DB_NAME = "activity_registration.db"

# 定義建立資料庫的函式
def initialize_database():

    conn = sqlite3.connect(DB_NAME)

    try:
        cursor = conn.cursor()

        # 每次執行都先刪除舊資料表
        cursor.execute("DROP TABLE IF EXISTS registrations")
        cursor.execute("DROP TABLE IF EXISTS activities")

        # 建立活動資料表
        cursor.execute("""
            CREATE TABLE activities (
                id INTEGER PRIMARY KEY AUTOINCREMENT,
                name TEXT NOT NULL,
                event_date TEXT NOT NULL,
                capacity INTEGER NOT NULL
            )
        """)

        # 建立報名紀錄資料表
        cursor.execute("""
            CREATE TABLE registrations (
                id INTEGER PRIMARY KEY AUTOINCREMENT,
                student_name TEXT NOT NULL,
                activity_id INTEGER NOT NULL,
                people INTEGER NOT NULL,
                status TEXT NOT NULL,
                FOREIGN KEY (activity_id) REFERENCES activities(id)
            )
        """)

        # 新增三場初始活動
        initial_data = [
            ("迎新晚會", "2026-11-20", 6),
            ("程式工作坊", "2026-11-25", 4),
            ("籃球友誼賽", "2026-11-28", 3)
        ]

        cursor.executemany("""
            INSERT INTO activities (name, event_date, capacity)
            VALUES (?, ?, ?)
        """, initial_data)

        conn.commit()

        print("資料庫建立成功！")
        print("已新增三場初始活動。")

    finally:
        conn.close()


# 執行資料庫初始化
initialize_database()
````

````python
# ============================================================
# 第 3 格：顯示資料表
# 提醒：定義 show_tables() 並呼叫一次
# ============================================================
# ============================================================
# 第 3 格：顯示資料表
# ============================================================

def show_tables():

    conn = sqlite3.connect(DB_NAME)

    try:

        print("===== 活動資料表 =====")

        activities_df = pd.read_sql_query(
            "SELECT * FROM activities",
            conn
        )

        display(activities_df)

        print("===== 報名紀錄資料表 =====")

        registrations_df = pd.read_sql_query(
            "SELECT * FROM registrations",
            conn
        )

        display(registrations_df)

    finally:
        conn.close()


# 顯示資料表
show_tables()
````

````python
# ============================================================
# 第 4 格：AI 解讀函式
# 提醒：回傳 (動作, 活動, 人數, 原始回覆)；不可直接相信 AI 的回覆
# ============================================================
# ============================================================
# 第 4 格：AI 解讀函式
# ============================================================

# AI 解讀結果暫存
ai_cache = {}

# 允許的動作
ALLOWED_ACTIONS = [
    "報名",
    "取消",
    "查詢",
    "統計"
]

# 允許的活動
ALLOWED_ACTIVITIES = [
    "迎新晚會",
    "程式工作坊",
    "籃球友誼賽"
]


def interpret_request(user_text):

    # 如果句子已經解讀過，直接使用暫存
    if user_text in ai_cache:
        return ai_cache[user_text]

    prompt = f"""
你是一個系學會活動報名系統的語意理解助手。

請分析使用者輸入的句子。

使用者輸入：
{user_text}

你只能輸出一行，格式如下：
動作,活動,人數

動作只能是：
報名、取消、查詢、統計、未知

活動只能是：
迎新晚會、程式工作坊、籃球友誼賽、未知

人數規則：
1. 沒有說人數時，預設為1。
2. 我和兩個朋友代表3人。
3. 我們三個人代表3人。
4. 我跟一個朋友代表2人。
5. 我和兩位朋友代表3人。
6. 請用阿拉伯數字輸出人數。
7. 即使人數是0或4，也要照實輸出，不可自行修正。

其他規則：
1. 查詢各活動報名狀況，動作為統計。
2. 無法判斷動作時，輸出未知。
3. 找不到活動時，活動輸出未知。
4. 不要解釋，不要輸出其他文字。

範例：
我要報名迎新晚會
報名,迎新晚會,1

我和兩個朋友報名程式工作坊
報名,程式工作坊,3

查詢各活動報名狀況
統計,未知,1

今天天氣很好
未知,未知,1
"""

    raw_reply = "AI 呼叫失敗"

    # 最多呼叫三次
    for attempt in range(3):

        try:

            response = client.chat.completions.create(
                model=MODEL,
                messages=[
                    {
                        "role": "system",
                        "content": "你是活動報名系統的語意分析助手。"
                    },
                    {
                        "role": "user",
                        "content": prompt
                    }
                ],
                temperature=0,
                reasoning_effort=REASONING_EFFORT
            )

            raw_reply = response.choices[0].message.content

            # 每次呼叫後等待兩秒
            time.sleep(2)

            break

        except Exception as e:

            error_message = str(e)

            print("AI 呼叫發生錯誤：", error_message)

            time.sleep(2)

            # 遇到額度限制或速率限制
            if "429" in error_message or "rate_limit" in error_message.lower():

                if attempt < 2:

                    print("遇到 429，等待 30 秒後重試。")
                    time.sleep(30)
                    continue

            break

    # --------------------------------------------------------
    # 驗證 AI 回覆，不直接相信 AI
    # --------------------------------------------------------

    reply = str(raw_reply).strip()

    # 清除可能出現的 Markdown 符號
    reply = reply.replace("```", "").strip()

    # 取得允許的動作
    action = "未知"

    for item in ALLOWED_ACTIONS:

        if item in reply:
            action = item
            break

    # 取得允許的活動
    activity = "未知"

    for item in ALLOWED_ACTIVITIES:

        if item in reply:
            activity = item
            break

    # 用正規表示式找出人數
    numbers = re.findall(r"\d+", reply)

    if numbers:

        people = int(numbers[-1])

    else:

        people = 1

    # 儲存結果，避免同一句話重複呼叫 AI
    result = (action, activity, people, raw_reply)

    ai_cache[user_text] = result

    return result
````

````python
# ============================================================
# 第 5 格：AI 解讀測試
# 提醒：8～10 句，要與規格書 7-1 一致
# ============================================================
# ============================================================
# 第 5 格：AI 解讀測試（共 10 句）
# ============================================================

ai_tests = [

    ("我要報名迎新晚會", "報名", "迎新晚會", 1),

    ("幫我取消程式工作坊", "取消", "程式工作坊", 1),

    ("查詢籃球友誼賽剩餘名額", "查詢", "籃球友誼賽", 1),

    ("我和兩個朋友要報名迎新晚會", "報名", "迎新晚會", 3),

    ("我們四個人要報名程式工作坊", "報名", "程式工作坊", 4),

    ("我要報名不存在的桌遊活動", "報名", "未知", 1),

    ("今天天氣很好", "未知", "未知", 1),

    ("幫我取消籃球友誼賽報名", "取消", "籃球友誼賽", 1),

    ("請查詢各活動報名狀況", "統計", "未知", 1),

    ("我要報名迎新晚會，0人", "報名", "迎新晚會", 0)

]


passed = 0

print("========== AI 解讀測試開始 ==========")

for i, test in enumerate(ai_tests):

    user_text, expected_action, expected_activity, expected_people = test

    # 呼叫 AI 解讀
    action, activity, people, raw_reply = interpret_request(user_text)

    # 判斷是否符合預期
    action_ok = (
        expected_action == "*"
        or action == expected_action
    )

    activity_ok = (
        expected_activity == "*"
        or activity == expected_activity
    )

    people_ok = (
        expected_people == "*"
        or people == expected_people
    )

    success = action_ok and activity_ok and people_ok

    if success:
        passed += 1

    print("\n第", i + 1, "句")
    print("使用者輸入：", user_text)
    print("AI 原始回覆：", raw_reply)
    print("系統解讀：", action, activity, people)
    print("測試結果：", "通過" if success else "未通過")


print("\n========== AI 測試總結 ==========")
print("總測試句數：", len(ai_tests))
print("通過句數：", passed)
print("未通過句數：", len(ai_tests) - passed)
````

````python
# ============================================================
# 第 6 格：功能函式
# 提醒：fetchone() 後先檢查 None；用 try/finally 關閉連線；統計用 JOIN＋GROUP BY
# ============================================================
# ============================================================
# 第 6 格：功能函式
# ============================================================


# ------------------------------------------------------------
# 功能一：查詢所有活動的統計資料
# ------------------------------------------------------------

def get_statistics():

    conn = sqlite3.connect(DB_NAME)

    try:

        cursor = conn.cursor()

        # 使用 LEFT JOIN 保留沒有報名紀錄的活動
        # 使用 GROUP BY 統計每場活動的報名人數

        cursor.execute("""
            SELECT
                a.name,
                a.event_date,
                a.capacity,
                COALESCE(SUM(r.people), 0),
                a.capacity - COALESCE(SUM(r.people), 0)

            FROM activities a

            LEFT JOIN registrations r
                ON a.id = r.activity_id
                AND r.status = ?

            GROUP BY
                a.id,
                a.name,
                a.event_date,
                a.capacity

            ORDER BY a.id
        """, ("已報名",))

        rows = cursor.fetchall()

        if not rows:
            return "目前沒有活動資料。"

        result = "========== 各活動報名統計 ==========\n"

        for row in rows:

            name = row[0]
            event_date = row[1]
            capacity = row[2]
            booked = row[3]
            remaining = row[4]

            result += (
                "\n活動：" + name +
                "\n日期：" + event_date +
                "\n總名額：" + str(capacity) +
                "\n已報名人數：" + str(booked) +
                "\n剩餘名額：" + str(remaining) +
                "\n--------------------------"
            )

        return result

    finally:
        conn.close()


# ------------------------------------------------------------
# 功能二：處理使用者需求
# ------------------------------------------------------------

def handle_request(user_text, student_name):

    # 先使用 AI 解讀
    action, activity, people, raw_reply = interpret_request(user_text)

    student_name = str(student_name).strip()

    # --------------------------------------------------------
    # 無法判斷動作
    # --------------------------------------------------------

    if action == "未知":

        conn = sqlite3.connect(DB_NAME)

        try:

            cursor = conn.cursor()

            cursor.execute("""
                SELECT name FROM activities
            """)

            rows = cursor.fetchall()

            names = [row[0] for row in rows]

            return (
                "無法判斷您的需求。\n"
                "目前提供的活動有：\n" +
                "\n".join(names)
            )

        finally:
            conn.close()

    # --------------------------------------------------------
    # 統計查詢
    # --------------------------------------------------------

    if action == "統計":

        return get_statistics()

    # --------------------------------------------------------
    # 檢查活動是否存在
    # --------------------------------------------------------

    if activity == "未知" or activity not in ALLOWED_ACTIVITIES:

        return "找不到您指定的活動，請重新輸入。"

    conn = sqlite3.connect(DB_NAME)

    try:

        cursor = conn.cursor()

        cursor.execute("""
            SELECT id, name, capacity
            FROM activities
            WHERE name = ?
        """, (activity,))

        activity_data = cursor.fetchone()

        # fetchone() 之後先檢查是否為 None
        if activity_data is None:

            return "找不到這場活動。"

        activity_id = activity_data[0]
        activity_name = activity_data[1]
        capacity = activity_data[2]

        # ----------------------------------------------------
        # 功能三：查詢剩餘名額
        # ----------------------------------------------------

        if action == "查詢":

            cursor.execute("""
                SELECT COALESCE(SUM(people), 0)
                FROM registrations
                WHERE activity_id = ?
                AND status = ?
            """, (activity_id, "已報名"))

            count_data = cursor.fetchone()

            if count_data is None:
                booked = 0
            else:
                booked = count_data[0]

            remaining = capacity - booked

            return (
                "活動：" + activity_name +
                "\n總名額：" + str(capacity) +
                "\n已報名：" + str(booked) +
                "\n剩餘名額：" + str(remaining)
            )

        # ----------------------------------------------------
        # 功能四：報名
        # ----------------------------------------------------

        if action == "報名":

            # 檢查姓名
            if not student_name:

                return "請輸入姓名才能報名。"

            # 檢查人數是否介於 1～3
            if people < 1 or people > 3:

                return "報名失敗：每次報名人數必須介於 1～3 人。"

            # 檢查是否已經報名同一個活動
            cursor.execute("""
                SELECT id
                FROM registrations
                WHERE student_name = ?
                AND activity_id = ?
                AND status = ?
            """, (student_name, activity_id, "已報名"))

            duplicate = cursor.fetchone()

            if duplicate is not None:

                return "報名失敗：您已經報名過這場活動。"

            # 檢查目前同時報名的活動數量
            cursor.execute("""
                SELECT COUNT(DISTINCT activity_id)
                FROM registrations
                WHERE student_name = ?
                AND status = ?
            """, (student_name, "已報名"))

            count_data = cursor.fetchone()

            if count_data is None:
                active_count = 0
            else:
                active_count = count_data[0]

            # 每個人最多同時報名兩個活動
            if active_count >= 2:

                return "報名失敗：每位學生最多同時報名 2 個活動。"

            # 計算目前已報名人數
            cursor.execute("""
                SELECT COALESCE(SUM(people), 0)
                FROM registrations
                WHERE activity_id = ?
                AND status = ?
            """, (activity_id, "已報名"))

            booked_data = cursor.fetchone()

            if booked_data is None:
                booked = 0
            else:
                booked = booked_data[0]

            remaining = capacity - booked

            # 檢查是否還有足夠名額
            if people > remaining:

                return (
                    "報名失敗：活動名額不足。\n"
                    "目前剩餘名額：" + str(remaining) + " 人"
                )

            # 新增報名紀錄
            cursor.execute("""
                INSERT INTO registrations
                (
                    student_name,
                    activity_id,
                    people,
                    status
                )
                VALUES (?, ?, ?, ?)
            """, (
                student_name,
                activity_id,
                people,
                "已報名"
            ))

            conn.commit()

            return (
                "報名成功！\n"
                "姓名：" + student_name +
                "\n活動：" + activity_name +
                "\n報名人數：" + str(people) + " 人" +
                "\n剩餘名額：" + str(remaining - people) + " 人"
            )

        # ----------------------------------------------------
        # 功能五：取消報名
        # ----------------------------------------------------

        if action == "取消":

            if not student_name:

                return "請輸入姓名才能取消報名。"

            # 查詢有效報名紀錄
            cursor.execute("""
                SELECT id, people
                FROM registrations
                WHERE student_name = ?
                AND activity_id = ?
                AND status = ?
                ORDER BY id DESC
                LIMIT 1
            """, (student_name, activity_id, "已報名"))

            registration = cursor.fetchone()

            if registration is None:

                return "取消失敗：查不到您的有效報名紀錄。"

            registration_id = registration[0]
            old_people = registration[1]

            # 將狀態改為已取消
            cursor.execute("""
                UPDATE registrations
                SET status = ?
                WHERE id = ?
            """, ("已取消", registration_id))

            conn.commit()

            return (
                "取消成功！\n"
                "姓名：" + student_name +
                "\n活動：" + activity_name +
                "\n已釋出名額：" + str(old_people) + " 人"
            )

        return "無法處理此項需求。"

    finally:

        conn.close()
````

````python
# ============================================================
# 第 7 格：主流程與流程測試
# 提醒：流程測試 8～10 步；最後呼叫 show_tables()
# ============================================================
# ============================================================
# 第 7 格：主流程與流程測試（共 9 步）
# ============================================================

# 重新建立資料庫，確保測試從初始狀態開始
initialize_database()


# 流程測試資料
flow_tests = [

    ("王小明", "我要報名迎新晚會"),

    ("王小明", "我要報名程式工作坊"),

    ("王小明", "我要報名籃球友誼賽"),

    ("陳小華", "我和兩個朋友要報名迎新晚會"),

    ("李小美", "我要報名迎新晚會"),

    ("王小明", "幫我取消程式工作坊"),

    ("王小明", "我要報名籃球友誼賽"),

    ("陳小華", "查詢迎新晚會剩餘名額"),

    ("幹部", "請查詢各活動報名狀況")

]


print("========== 流程測試開始 ==========")

for i, test in enumerate(flow_tests):

    student_name = test[0]
    user_text = test[1]

    print("\n========== 第", i + 1, "步 ==========")

    print("姓名：", student_name)
    print("使用者輸入：", user_text)

    # 執行系統功能
    result = handle_request(user_text, student_name)

    print("系統回應：")
    print(result)


print("\n========== 流程測試完成 ==========")


# 最後顯示資料表
print("\n========== 最終資料庫內容 ==========")

show_tables()
````
